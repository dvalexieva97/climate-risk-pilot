# Validate Portugal pilot outputs

Run after `climate-risk run-pilot --config config/pilot_portugal.yaml`.

```python
from pathlib import Path
import json
import rasterio
import pandas as pd

ROOT = Path('..').resolve()
processed = ROOT / 'data/processed/portugal'
events = ROOT / 'data/events/portugal'
manifest = json.loads((processed / 'manifest.json').read_text())
manifest
```

```python
for hazard, paths in manifest['layers'].items():
    hazard_path = Path(paths['hazard'])
    with rasterio.open(hazard_path) as src:
        data = src.read(1, masked=True)
        print(hazard, hazard_path.name, 'min', float(data.min()), 'max', float(data.max()))
    summary = events / hazard / f"event_summary_{manifest['event_history']['years']}y.json"
    if summary.exists():
        print('  events', json.loads(summary.read_text()))
```
